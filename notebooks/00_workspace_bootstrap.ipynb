# Workspace Bootstrap

Load a reusable `AnalysisWorkspace` that every other notebook can build on.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'analysis_module').is_dir():
            return candidate
    return start

ROOT = find_repo_root(Path.cwd())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

ROOT
```

```python
from analysis_module import load_workspace, lens_coverage

ws = load_workspace(
    scores_path=ROOT / 'scores.json',
    lenses_path=ROOT / 'lenses',
    aggregation='latest',
)

print('items:', len(ws.item_ids))
print('lenses:', len(ws.lens_names))
print('aggregation:', ws.aggregation)
print('coverage:', lens_coverage(ws, normalized=False))
```
