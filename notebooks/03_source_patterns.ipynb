# Source Patterns

Look for differences in lens profiles by source and quick nearest-source checks.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'analysis_module').is_dir():
            return candidate
    return start

ROOT = find_repo_root(Path.cwd())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from analysis_module import load_workspace, source_lens_means, complete_item_rows

ws = load_workspace(ROOT / 'scores.json', ROOT / 'lenses', aggregation='latest')
source_means = source_lens_means(ws, normalized=True)
source_df = pd.DataFrame(source_means).T
source_df
```

```python
source_df['mean_across_lenses'] = source_df.mean(axis=1)
source_df.sort_values('mean_across_lenses', ascending=False)
```

```python
item_ids, source_labels, matrix = complete_item_rows(ws, normalized=True)
print('complete rows:', len(item_ids))
print('sources:', sorted(set(source_labels)))
```
