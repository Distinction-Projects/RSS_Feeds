# Lens Matrix Playground

Explore article-level lens scores as a table and rank stories by any lens.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'analysis_module').is_dir():
            return candidate
    return start

ROOT = find_repo_root(Path.cwd())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from analysis_module import load_workspace, to_pandas_articles

ws = load_workspace(ROOT / 'scores.json', ROOT / 'lenses', aggregation='latest')
df = to_pandas_articles(ws, normalized=True, include_incomplete=False)
df.head()
```

```python
lens_cols = ws.lens_names
df[lens_cols].describe().T.sort_values('mean', ascending=False)
```

```python
lens_name = ws.lens_names[0]
df[['item_id', 'source', 'title', lens_name]].sort_values(lens_name, ascending=False).head(10)
```
