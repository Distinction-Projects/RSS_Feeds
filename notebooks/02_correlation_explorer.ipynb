# Correlation Explorer

Inspect pairwise lens correlations/covariances and find strongest relationships.

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'analysis_module').is_dir():
            return candidate
    return start

ROOT = find_repo_root(Path.cwd())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from analysis_module import load_workspace, pairwise_metric_matrix

ws = load_workspace(ROOT / 'scores.json', ROOT / 'lenses', aggregation='latest')
corr, counts = pairwise_metric_matrix(ws, metric='correlation', normalized=True)
corr_df = pd.DataFrame(corr, index=ws.lens_names, columns=ws.lens_names)
count_df = pd.DataFrame(counts, index=ws.lens_names, columns=ws.lens_names)
corr_df
```

```python
pairs = []
for i, lens_a in enumerate(ws.lens_names):
    for j, lens_b in enumerate(ws.lens_names):
        if j <= i:
            continue
        value = corr[i][j]
        n = counts[i][j]
        if value is None:
            continue
        pairs.append({'lens_a': lens_a, 'lens_b': lens_b, 'corr': value, 'n': n})

pair_df = pd.DataFrame(pairs).sort_values('corr', ascending=False)
pair_df.head(10)
```
